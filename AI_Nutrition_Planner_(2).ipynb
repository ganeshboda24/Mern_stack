````python
import os
import subprocess
import time
from pyngrok import ngrok, conf

# Ensure ngrok auth token is set (it should be from previous steps)
conf.get_default().auth_token = os.environ.get("NGROK_AUTH_TOKEN")

print('Stopping any existing ngrok tunnels and Flask processes...')
ngrok.kill() # Kill all ngrok tunnels

# Attempt to kill any previously started Flask processes specifically for 'Nourix/app.py'
# This uses 'pkill -f' which is a common way to kill processes by command line in Linux environments.
# This might not always be perfect but is a robust attempt for Colab.
os.system("pkill -f 'python Nourix/app.py'")
time.sleep(2) # Give some time for processes to terminate gracefully

print('🚀 Starting Nourix Flask application in the background...')
# Run Flask in a separate process in the background
# We redirect output to files to prevent blocking the Colab cell
# and allow subsequent cells (like ngrok) to run.
# nohup ensures the process continues if the shell exits, and '&' backgrounds it.
# This also captures stdout/stderr to files for debugging if needed.
flask_log_path = "./flask_output.log"
with open(flask_log_path, "w") as log_file:
    flask_process = subprocess.Popen(
        ["nohup", "python", "Nourix/app.py"], # <--- Changed to run Nourix/app.py
        stdout=log_file,
        stderr=log_file,
        preexec_fn=os.setsid # Detach from parent process group
    )
print(f"Flask app background process started. Output redirected to {flask_log_path}. PID: {flask_process.pid}")

# Give Flask a moment to start up on port 5000
time.sleep(5)

print("Connecting ngrok tunnel...")
# Create a new tunnel to Flask (port 5000)
# Retry logic in case Flask isn't immediately ready
public_url = None
for _ in range(5):
    try:
        public_url = ngrok.connect(5000, bind_tls=True) # Ensure HTTPS
        break
    except Exception as e:
        print(f"Ngrok connection failed: {e}. Retrying...")
        time.sleep(5)

if public_url:
    print(f"🎉 Your Nourix app is live at: {public_url.public_url}")
    print("Note: The Flask server is running in the background. To check its logs, view './flask_output.log'.")
    print("If you need to stop it explicitly (without restarting runtime), you might need to use its PID or `os.system(\"pkill -f 'python Nourix/app.py'\")`.")
else:
    print("❌ Failed to establish ngrok tunnel after several attempts. Check flask_output.log for errors.")
````

````python
app_py_content = '''
import os
import json
import traceback
from flask import Flask, request, jsonify, render_template
from groq import Groq

app = Flask(__name__)

# Initialize Groq client
GROQ_API_KEY = os.environ.get("GROQ_API_KEY")
if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY environment variable not set.")

groq_client = Groq(api_key=GROQ_API_KEY)

# --- Routes ---

@app.route("/")
def home():
    return render_template("home.html")

@app.route("/planner")
def planner():
    return render_template("planner.html")

@app.route("/health")
def health():
    return render_template("health.html")

@app.route("/about")
def about():
    return render_template("about.html")

@app.route("/generate", methods=["POST"])
def generate_meal_plan():
    try:
        data = request.get_json()
        if not data:
            return jsonify({"error": "Invalid JSON data provided."}), 400

        # Extract user inputs
        breakfast = data.get("breakfast", "").strip()
        lunch = data.get("lunch", "").strip()
        dinner = data.get("dinner", "").strip()
        cuisine = data.get("cuisine", "").strip()
        diet_preference = data.get("dietPreference", "").strip()
        fitness_goal = data.get("fitnessGoal", "").strip()
        allergies = data.get("allergies", "").strip()
        num_servings = data.get("numServings", "").strip()
        cooking_time = data.get("cookingTime", "").strip()

        # Basic validation
        if not all([breakfast, lunch, dinner, cuisine, diet_preference, fitness_goal, num_servings, cooking_time]):
            return jsonify({"error": "All meal fields, cuisine, diet preference, fitness goal, number of servings, and cooking time are required."}), 400

        prompt = f"""
        You are Nourix, an expert AI Nutrition Planner. Generate a simple and beginner-friendly nutrition plan.

        The user will provide:
        - Breakfast: {breakfast}
        - Lunch: {lunch}
        - Dinner: {dinner}
        - Cuisine: {cuisine}
        - Goal: {fitness_goal}

        Return the response in the following JSON format only:
        {{
            "meal_plan": {{
                "breakfast": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": ""
                }},
                "lunch": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": ""
                }},
                "dinner": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": ""
                }}
            }},
            "ingredients_table": [
                {{
                    "ingredient": "",
                    "quantity": "",
                    "benefit": ""
                }}
            ],
            "cooking_steps": [
                ""
            ],
            "ai_nutrition_tips": [
                ""
            ],
            "health_benefits": [
                ""
            ],
            "daily_nutrition_summary": {{
                "calories": "",
                "protein": "",
                "carbohydrates": "",
                "fat": "",
                "fiber": ""
            }}
        }}
        Include 8-10 ingredients in the ingredients_table. Keep the cooking steps simple (5 steps).
        Provide exactly 5 short tips for ai_nutrition_tips.
        List 5 simple benefits for health_benefits.
        Ensure all values are realistic and well-formatted.
        """

        chat_completion = groq_client.chat.completions.create(
            messages=[
                {
                    "role": "user",
                    "content": prompt,
                }
            ],
            model="llama-3.3-70b-versatile", # Updated model name
            response_format={"type": "json_object"},
            temperature=0.7,
            max_tokens=2000 # Adjust as needed
        )

        response_content = chat_completion.choices[0].message.content
        meal_plan_data = json.loads(response_content)

        return jsonify(meal_plan_data)

    except json.JSONDecodeError:
        return jsonify({"error": "AI response was not a valid JSON object. Please try again."}), 500
    except Exception as e:
        app.logger.error(f"Error generating meal plan: {e}")
        app.logger.error(traceback.format_exc())
        return jsonify({"error": f"An unexpected error occurred: {str(e)}"}), 500

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000, debug=True)
'''

with open("Nourix/app.py", "w") as f:
    f.write(app_py_content)

print("✅ app.py updated in Nourix/app.py with new AI prompt and JSON structure")
````

````python
import os
import subprocess
import time
from pyngrok import ngrok, conf

# Ensure ngrok auth token is set (it should be from previous steps)
conf.get_default().auth_token = os.environ.get("NGROK_AUTH_TOKEN")

print('Stopping any existing ngrok tunnels and Flask processes...')
ngrok.kill() # Kill all ngrok tunnels

# Attempt to kill any previously started Flask processes specifically for 'Nourix/app.py'
# This uses 'pkill -f' which is a common way to kill processes by command line in Linux environments.
# This might not always be perfect but is a robust attempt for Colab.
os.system("pkill -f 'python Nourix/app.py'")
time.sleep(2) # Give some time for processes to terminate gracefully

print('🚀 Starting Nourix Flask application in the background...')
# Run Flask in a separate process in the background
# We redirect output to files to prevent blocking the Colab cell
# and allow subsequent cells (like ngrok) to run.
# nohup ensures the process continues if the shell exits, and '&' backgrounds it.
# This also captures stdout/stderr to files for debugging if needed.
flask_log_path = "./flask_output.log"
with open(flask_log_path, "w") as log_file:
    flask_process = subprocess.Popen(
        ["nohup", "python", "Nourix/app.py"], # Using nohup to ensure it runs independently
        stdout=log_file,
        stderr=log_file,
        preexec_fn=os.setsid # Detach from parent process group
    )
print(f"Flask app background process started. Output redirected to {flask_log_path}. PID: {flask_process.pid}")

# Give Flask a moment to start up on port 5000
time.sleep(5)

print("Connecting ngrok tunnel...")
# Create a new tunnel to Flask (port 5000)
# Retry logic in case Flask isn't immediately ready
public_url = None
for _ in range(5):
    try:
        public_url = ngrok.connect(5000, bind_tls=True) # Ensure HTTPS
        break
    except Exception as e:
        print(f"Ngrok connection failed: {e}. Retrying...")
        time.sleep(5)

if public_url:
    print(f"🎉 Your Nourix app is live at: {public_url.public_url}")
    print("Note: The Flask server is running in the background. To check its logs, view './flask_output.log'.")
    print("If you need to stop it explicitly (without restarting runtime), you might need to use its PID or `os.system(\"pkill -f 'python Nourix/app.py'\")`.")
else:
    print("❌ Failed to establish ngrok tunnel after several attempts. Check flask_output.log for errors.")
````

````python
app_py_content_updated = '''
import os
import json
import traceback
import requests # Import requests for Pixabay API
from flask import Flask, request, jsonify, render_template
from groq import Groq

# Explicitly set the template and static folders relative to the app.py file
app = Flask(__name__,
            template_folder=os.path.join(os.path.dirname(__file__), 'templates'),
            static_folder=os.path.join(os.path.dirname(__file__), 'static'))

# Initialize Groq client
GROQ_API_KEY = os.environ.get("GROQ_API_KEY")
PIXABAY_API_KEY = os.environ.get("PIXABAY_API_KEY") # Get Pixabay API key

if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY environment variable not set.")

groq_client = Groq(api_key=GROQ_API_KEY)

# --- Helper function for fetching image from Pixabay ---
def get_image_url(query):
    if not PIXABAY_API_KEY:
        print("Warning: PIXABAY_API_KEY not set. Using default image.")
        return "https://images.unsplash.com/photo-1540189549336-e619d45e0ace?q=80&w=2670&auto=format&fit=crop&ixlib=rb-4.0.3&ixid=M3wxMjA3fDB8MHxwaG90by1wYWdlfHx8fGVufDB8fHx8fA%3D%3D" # Default image

    search_url = f"https://pixabay.com/api/?key={PIXABAY_API_KEY}&q={query}&image_type=photo&per_page=3&safesearch=true"
    try:
        response = requests.get(search_url)
        response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
        data = response.json()
        if data['hits']:
            return data['hits'][0]['webformatURL'] # Return URL of the first image
    except requests.exceptions.RequestException as e:
        app.logger.error(f"Error fetching image from Pixabay for query '{query}': {e}")
    except Exception as e:
        app.logger.error(f"Unexpected error processing Pixabay response: {e}")

    return "https://images.unsplash.com/photo-1540189549336-e619d45e0ace?q=80&w=2670&auto=format&fit=crop&ixlib=rb-4.0.3&ixid=M3wxMjA3fDB8MHxwaG90by1wYWdlfHx8fGVufDB8fHx8fA%3D%3D" # Default image if API fails


# --- Routes ---

@app.route("/")
def home():
    return render_template("home.html")

@app.route("/planner")
def planner():
    return render_template("planner.html")

@app.route("/health")
def health():
    return render_template("health.html")

@app.route("/about")
def about():
    return render_template("about.html")

@app.route("/generate", methods=["POST"])
def generate_meal_plan():
    try:
        data = request.get_json()
        if not data:
            return jsonify({"error": "Invalid JSON data provided."}), 400

        # Extract user inputs
        breakfast = data.get("breakfast", "").strip()
        lunch = data.get("lunch", "").strip()
        dinner = data.get("dinner", "").strip()
        cuisine = data.get("cuisine", "").strip()
        diet_preference = data.get("dietPreference", "").strip()
        fitness_goal = data.get("fitnessGoal", "").strip()
        allergies = data.get("allergies", "").strip()
        num_servings = data.get("numServings", "").strip()
        cooking_time = data.get("cookingTime", "").strip()

        # Basic validation
        if not all([breakfast, lunch, dinner, cuisine, diet_preference, fitness_goal, num_servings, cooking_time]):
            return jsonify({"error": "All meal fields, cuisine, diet preference, fitness goal, number of servings, and cooking time are required."}), 400

        prompt = f"""
        You are Nourix, an expert AI Nutrition Planner. Generate a simple and beginner-friendly nutrition plan.

        The user will provide:
        - Breakfast: {breakfast}
        - Lunch: {lunch}
        - Dinner: {dinner}
        - Cuisine: {cuisine}
        - Goal: {fitness_goal}
        - Dietary Preference: {diet_preference}
        - Allergies: {allergies if allergies else 'None'}
        - Number of Servings: {num_servings}
        - Max Cooking Time: {cooking_time} minutes

        Return the response in the following JSON format only:
        {{
            "meal_plan": {{
                "breakfast": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": "",
                    "image_query": "" // Dish name only, e.g., 'Oatmeal with berries'
                }},
                "lunch": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": "",
                    "image_query": "" // Dish name only
                }},
                "dinner": {{
                    "meal_name": "",
                    "calories": "",
                    "protein": "",
                    "carbs": "",
                    "fat": "",
                    "image_query": "" // Dish name only
                }}
            }},
            "ingredients_table": [
                {{
                    "ingredient": "",
                    "quantity": "",
                    "benefit": ""
                }}
            ],
            "cooking_steps": [
                ""
            ],
            "ai_nutrition_tips": [
                ""
            ],
            "health_benefits": [
                ""
            ],
            "daily_nutrition_summary": {{
                "calories": "",
                "protein": "",
                "carbohydrates": "",
                "fat": "",
                "fiber": ""
            }}
        }}
        Include 8-10 ingredients in the ingredients_table. Keep the cooking steps simple (5 steps).
        Provide exactly 5 short tips for ai_nutrition_tips.
        List 5 simple benefits for health_benefits.
        Ensure all values are realistic and well-formatted. Make sure 'image_query' only contains the name of the dish.
        """

        chat_completion = groq_client.chat.completions.create(
            messages=[
                {
                    "role": "user",
                    "content": prompt,
                }
            ],
            model="llama-3.3-70b-versatile", # Updated model name
            response_format={"type": "json_object"},
            temperature=0.7,
            max_tokens=2000 # Adjust as needed
        )

        response_content = chat_completion.choices[0].message.content
        meal_plan_data = json.loads(response_content)

        # Fetch images for each meal
        if "meal_plan" in meal_plan_data:
            for meal_type in ["breakfast", "lunch", "dinner"]:
                if meal_type in meal_plan_data["meal_plan"] and "image_query" in meal_plan_data["meal_plan"][meal_type]:
                    query = meal_plan_data["meal_plan"][meal_type]["image_query"]
                    meal_plan_data["meal_plan"][meal_type]["image_url"] = get_image_url(query)

        return jsonify(meal_plan_data)

    except json.JSONDecodeError:
        app.logger.error(f"AI response was not a valid JSON object: {response_content}")
        return jsonify({"error": "AI response was not a valid JSON object. Please try again."}), 500
    except Exception as e:
        app.logger.error(f"Error generating meal plan: {e}")
        app.logger.error(traceback.format_exc())
        return jsonify({"error": f"An unexpected error occurred: {str(e)}"}), 500

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000, debug=False)
'''

with open("Nourix/app.py", "w") as f:
    f.write(app_py_content_updated)

print("✅ Nourix/app.py updated with enhanced AI prompt, Pixabay integration, and JSON structure")
````
